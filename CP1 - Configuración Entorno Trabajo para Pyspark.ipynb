# 📂 Configurar Repositorio de Código

## 🔹 Crear un nuevo repositorio en GitHub Desktop
1. Abrir **VS Code**.  
2. Abrir la aplicación **GitHub Desktop** e iniciar sesión con vuestra cuenta.  
3. Ir a **Archivo → Nuevo Repositorio**.  
4. Configurar:  
   - **Nombre**: `fundamentos_big_data`  
   - **Descripción**: Repositorio “Fundamentos de Big Data”  
   - **Local Path**: normalmente en `Documentos\GitHub\fundamentos_big_data`  
5. Pulsar en **Crear el Repositorio**.  

---

## 🔹 Publicar Repositorio en GitHub
1. Publicar el repositorio con:  
   - **Nombre**: `fundamentos_big_data`  
   - **Descripción**: Repositorio “Fundamentos de Big Data”  
   - Marcar ✅ **Mantener el repositorio privado**  
2. Ir al menú **Branch** y crear una nueva rama llamada `develop`.  
3. Pulsar en **Publicar Branch** → se cambia directamente a esa rama.  

---

## 🔹 Añadir archivo inicial
1. Pulsar en **Mostrar en Carpeta**.  
2. Copiar dentro el archivo `spark.ipynb`.  
3. Volver a **GitHub Desktop** → ahora aparecerá el archivo nuevo.  
4. En la parte inferior hacer un **Commit** con:  
   - **Título**: `Subir archivo de trabajo`  
   - **Descripción**: `Archivo Jupyter para CP1`  
5. Pulsar en **Commit**.  
6. Luego pulsar en **Push** (subir cambios).

````python
ventas_df = spark.createDataFrame(ventas_data)
clientes_df = spark.createDataFrame(clientes_data)

ventas_df.printSchema(); clientes_df.printSchema()
````

````python
ventas_df.show(5,0)
````

````python
clientes_df.show() 
````

````python
agg_df = (
    ventas_df
      .groupBy("ciudad", "categoria")
      .agg(
          F.round(F.sum("importe"), 2).alias("importe_total"),
          F.round(F.avg("importe"), 2).alias("ticket_medio"),
          F.countDistinct("cliente_id").alias("n_clientes")
      )
)
agg_df.orderBy("ciudad", "categoria").show()
#agg_df.explain(True)
````

````python
joined_df = ventas_df.join(F.broadcast(clientes_df), on="cliente_id", how="left")
joined_df.select("pedido_id", "nombre", "ciudad", "categoria", "importe", "edad").show()
#joined_df.explain(True)
````
